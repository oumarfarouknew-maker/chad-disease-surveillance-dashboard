```python
## Exploratory Data Analysis (EDA) 
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the cleaned health dataset
df = pd.read_csv("health_surveillance_cleaned.csv")

# Display the first five rows
display(df.head())
```

```python
# Dataset dimensions
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Column names and data types
print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

# Statistical summary
display(df.describe(include="all").T)
```

```python
missing = df.isnull().sum()

missing_summary = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": (missing / len(df) * 100).round(2)
})

display(missing_summary.sort_values(
    "Missing_Percentage",
    ascending=False
))
```

```python
missing_summary["Missing_Percentage"].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Missing Values by Column")
plt.xlabel("Variables")
plt.ylabel("Missing Values (%)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
if "Disease" in df.columns:
    disease_counts = df["Disease"].value_counts()

    display(disease_counts)

    disease_counts.plot(
        kind="bar",
        figsize=(10, 5)
    )

    plt.title("Distribution of Health Cases by Disease")
    plt.xlabel("Disease")
    plt.ylabel("Number of Records")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
if "Region" in df.columns:
    region_counts = df["Region"].value_counts()

    display(region_counts)

    region_counts.plot(
        kind="bar",
        figsize=(10, 5)
    )

    plt.title("Health Records by Region")
    plt.xlabel("Region")
    plt.ylabel("Number of Records")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
if "Age" in df.columns:
    df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

    display(df["Age"].describe())

    df["Age"].dropna().plot(
        kind="hist",
        bins=15,
        figsize=(9, 5)
    )

    plt.title("Age Distribution of Recorded Cases")
    plt.xlabel("Age in years")
    plt.ylabel("Number of Records")
    plt.tight_layout()
    plt.show()
```

```python
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

    monthly_cases = (
        df.dropna(subset=["Date"])
          .set_index("Date")
          .resample("MS")
          .size()
    )

    monthly_cases.plot(
        kind="line",
        marker="o",
        figsize=(10, 5)
    )

    plt.title("Monthly Health Records Over Time")
    plt.xlabel("Month")
    plt.ylabel("Number of Records")
    plt.tight_layout()
    plt.show()
```

```python
# Save descriptive statistics
df.describe(include="all").T.to_csv(
    "health_eda_summary.csv",
    encoding="utf-8-sig"
)

print("EDA summary saved successfully.")
```

```python
## 5/Epidemiological Data Analysis
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("health_surveillance_cleaned.csv")

print("Dataset size:", df.shape)
print("Columns:", df.columns.tolist())

display(df.head())
```

```python
disease_summary = (
    df.groupby("disease")
      .size()
      .reset_index(name="Reported_Cases")
      .sort_values("Reported_Cases", ascending=False)
)

display(disease_summary)
```

```python
disease_summary.plot(
    x="disease",
    y="Reported_Cases",
    kind="bar",
    legend=False,
    figsize=(10, 5)
)

plt.title("Reported Cases by disease")
plt.xlabel("disease")
plt.ylabel("Number of Records")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
region_summary = (
    df.groupby("region")
      .size()
      .reset_index(name="Reported_Cases")
      .sort_values("Reported_Cases", ascending=False)
)

display(region_summary)
```

```python
region_summary.plot(
    x="region",
    y="Reported_Cases",
    kind="bar",
    legend=False,
    figsize=(10, 5)
)

plt.title("Reported Cases by region")
plt.xlabel("Region")
plt.ylabel("Number of Records")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
df["report_date"] = pd.to_datetime(df["report_date"], errors="coerce")

monthly_cases = (
    df.dropna(subset=["report_date"])
      .groupby(df["report_date"].dt.to_period("M"))
      .size()
)

monthly_cases.index = monthly_cases.index.to_timestamp()

display(monthly_cases)
```

```python
monthly_cases.plot(
    kind="line",
    marker="o",
    figsize=(10, 5)
)

plt.title("Monthly Reported Cases")
plt.xlabel("Month")
plt.ylabel("Number of Records")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Illustrative example only
new_cases = 240
population = 120_000

incidence_per_100k = (
    new_cases / population
) * 100_000

print(f"Incidence: {incidence_per_100k:.1f} per 100,000")
```

```python
# Illustrative example only
total_cases = 500
deaths = 15

cfr = (deaths / total_cases) * 100

print(f"Case fatality ratio: {cfr:.2f}%")
```

```python
epi_table = pd.crosstab(
    df["region"],
    df["disease"],
    margins=True,
    margins_name="Total"
)

display(epi_table)
```

```python
disease_summary.to_csv(
    "epidemiological_disease_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

region_summary.to_csv(
    "epidemiological_region_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Epidemiological summaries saved successfully.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the full health surveillance dataset
df = pd.read_csv("health_surveillance_cleaned.csv")

# Count records by disease
disease_summary = (
    df.groupby("disease")
      .size()
      .reset_index(name="Reported_Cases")
      .sort_values("Reported_Cases", ascending=False)
)

# Display the results
display(disease_summary)
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    disease_summary["disease"],
    disease_summary["Reported_Cases"]
)

plt.title(
    "Reported Health Cases by Disease",
    fontsize=14
)

plt.xlabel("Disease")
plt.ylabel("Number of Records")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
region_summary = (
    df.groupby("region")
      .size()
      .reset_index(name="Reported_Cases")
      .sort_values("Reported_Cases", ascending=False)
)

plt.figure(figsize=(10, 6))

plt.bar(
    region_summary["region"],
    region_summary["Reported_Cases"]
)

plt.title("Reported Health Records by Region", fontsize=14)
plt.xlabel("Region")
plt.ylabel("Number of Records")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
df["report_date"] = pd.to_datetime(df["report_date"], errors="coerce")

monthly_cases = (
    df.dropna(subset=["report_date"])
      .groupby(df["report_date"].dt.to_period("M"))
      .size()
)

monthly_cases.index = monthly_cases.index.to_timestamp()

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_cases.index,
    monthly_cases.values,
    marker="o"
)

plt.title("Monthly Reported Health Records")
plt.xlabel("Month")
plt.ylabel("Number of Records")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
